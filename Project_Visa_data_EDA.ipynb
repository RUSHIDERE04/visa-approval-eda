# Data Visualization

**Univariate Analysis**

$VisaStatusDistributions(Case Status)$

```python
# Case Status Distributions
visa_df['case_status'].value_counts()
```

```python
plt.figure(figsize=(7,4))
plt.title('Visa Status Distribution')
visa_df['case_status'].value_counts().plot(kind='bar')
plt.xlabel('Case Status')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

$EducationLevelDistributions$

```python
plt.figure(figsize=(7,4))
plt.title('Education Level Distribution')
visa_df['education_of_employee'].value_counts().plot(kind='bar')
plt.xlabel('Education Level Distribution')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

$Full Time Position Distribution$

```python
plt.figure(figsize=(6,5))
plt.title('Full Time Position Distribution')
sns.countplot(data=visa_df, x='full_time_position')
plt.xlabel('Full Time Position Distribution')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

$Job Experienece Distribution$

```python
plt.figure(figsize=(4,5))
plt.title('Job Experienece Distribution')
visa_df['has_job_experience'].value_counts().plot(kind='bar')
plt.xlabel('Job Experience')
plt.xticks(rotation=0)
plt.ylabel('Count')
plt.show()
```

## BiVariate Analysis

**Education VS Visa Status**  (Case Status)

```python
pd.crosstab(visa_df['education_of_employee'],visa_df['case_status'], normalize='index').plot(kind='bar', figsize=(7,4))
plt.title('Education VS Visa Status')
plt.xlabel('Education Of Employee')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

**Job Experience Vs Visa Status**  (Case Status)

```python
plt.figure(figsize=(4,5))
plt.title('Job Experience')
sns.countplot(data=visa_df, x='has_job_experience', hue='case_status')
plt.xlabel('Job Experience')
plt.ylabel('Count')
plt.show()
```

**No Of Employees Vs Visa Status**  (Case Status)

```python
plt.figure(figsize=(5,4))
plt.title('No Of Employees Vs Visa Status')
sns.stripplot(data=visa_df, x='no_of_employees', hue='case_status')
plt.xlabel('No Of Employees')
plt.ylabel('Count')
plt.show()
```

**Prevailing Wage Vs Visa Status** (Case Status)

```python
plt.figure(figsize=(4,4))
plt.title('Prevailing Wage Vs Visa Status')
sns.stripplot(data=visa_df, x='prevailing_wage', hue='case_status')
plt.xlabel('Prevailing Wage')
plt.ylabel('Count')
plt.show()
```

## Geographical Analysis

**Region Of Employment Analysis**

```python
plt.figure(figsize=(7,4))
plt.title('Region Of Employment Analysis')
sns.countplot(data=visa_df, x='region_of_employment', hue='case_status')
plt.xlabel('Region Of Employment')
plt.ylabel('Count')
plt.show()
```

## Feature Engineering

```python
# Create A New Column Age Of Company
visa_df['age_of_company']=2026 - visa_df['yr_of_estab']
```

```python
plt.figure(figsize=(7,4))
plt.title('Age Of Company Vs Case Status')
sns.histplot(data=visa_df, x='age_of_company', hue='case_status')
plt.xlabel('Age Of Company')
plt.ylabel('Case Status')
plt.show()
```

## Correlation Analysis

**Correlation Analysis with Numerical columns**

```python
num_col=['no_of_employees', 'age_of_company', 'prevailing_wage']
sns.heatmap(visa_df[num_col].corr(), annot=True, cmap=None)
plt.title('Correlation Matrix Heat Map')
plt.show()
```
